Problem 6:

Take an integer N. Print all numbers from 1 to N, but skip multiples of 3 using continue. Sample Input: 10 Sample Output: 1 2 4 5 7 8 10

```python
N = int(input())

for i in range(1, N + 1):
    if i % 3 == 0:
        continue
    print(i,end=" ")
```

Problem 7

Keep taking integers as input. Add them to a total. Stop when a negative number appears (break). Do not add that negative number. Then print the total. Sample Input: 4 8 15 -3 23 Sample Output: 27

```python
sum=0
for i in range(-1000,1000):
    num=int(input("Enter Integer:"))
    if num>0:
        sum=sum+num
    else:
        break

print(sum)
```

Problem 8:

Take a word as input. Print the first character, the last character, and the length, separated by spaces. Sample Input: Python Sample Output: P n 6

```python
word=input("Enter a word: (eg. Hello):")
processed_word=word.lower()
first=processed_word[0]
last=processed_word[-1]
length=len(processed_word)
print(f"{first} {last} {length}")
```

Problem 9

Take a string as input. Count how many vowels it has (a, e, i, o, u in both lowercase and uppercase) using a loop. Print only the count. Sample Input: Education Sample Output: 5

```python
s = input()
count = 0

for ch in s:
    if ch in "aeiouAEIOU":
        count += 1

print(count)
```
